```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/rishirokde-official/Python-for-Data-Science/refs/heads/main/python-for-data-science-main/05.Data%20Extraction/players_20.csv")
```

```python
df
```

```python
df=df[['long_name','age','dob','height_cm','weight_kg','nationality','club']]
```

```python
df
```

Drop Row

```python
#drop row (delete)
df.drop('L. Messi' , axis=0)
#all the rows of messi are deleated
df.drop(index=['L. Messi'])
```

```python
#drop two or more rows and update data (inplace=True)
df.drop(['L. Messi','Neymar Jr'],axis=0 , inplace=True)
```

```python
df
```

Drop Columns

```python
#drop one column
df.drop('long_name',axis=1)
```

```python
#drop column by its position (last column deleated)
df.drop(df.columns[[-1]],axis=1)
```

```python
# drop two or more columns and update data (inplace=True)
df.drop(['long_name' , 'dob'],axis=1 , inplace=True)
```
