```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/rishirokde-official/Python-for-Data-Science/refs/heads/main/python-for-data-science-main/05.Data%20Extraction/players_20.csv")
```

```python
df
```

```python
df.set_index('short_name' , inplace=True)
```

```python
df
```

```python
df=df[['long_name','age','dob','height_cm','weight_kg','nationality' , 'club']].copy()
```

```python
df
```

sample()

```python
# extract 10 random elements from the "nationality" column
df['nationality'].sample(10,random_state=99)
```

```python
#extract 20% data from table
df.sample(frac=0.2,random_state=64)
```

```python
df.sample(frac=2, replace=True, random_state=99)
```

query()

```python
#select player order than 34
df.query("age>34")
```

```python
# select players older than 34 from Italy
df.query("age>34 and nationality=='Italy'")
```

```python
df.query("not(age>34)")
```

```python
#convert height to metres and select those with height above 1.8
df.query("height_cm/100 > 1.8")
```

```python
# select players that were born after 1990

# check out data types
df.dtypes
# conver "dob" column to datetime type
df['dob'] = df['dob'].astype('datetime64[ns]')
# query
df.query("dob.dt.year > 1990")

# exercise: write the equivalent boolean slicing
df[df['dob'].dt.year>1990]
```
