selecting with single value
loc[row_lable , column_lable ]


```python
#get all data of L.Messi
df.loc['L. Messi']
```

```python
#get height of messi
df.loc['L. Messi','height_cm']
```

```python
#get the weight of cristino ronaldo
df.loc['Cristiano Ronaldo','weight_kg']
```

```python
#get all rows inside the 'heiight_cmcolumn
df.loc[:,'height_cm']
```

```python
#get particular info of one player
df.loc['L. Messi',:]
```

selecting with list of values

```python
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'weight_kg']
```

```python
# get the height and weight of L.Messi
df.loc['L. Messi', ['height_cm', 'weight_kg']]
```

selecting a range of data with a slice
df.loc[players, 'age':'club']

```python
#slice colum lables
players=['L. Messi','Cristiano Ronaldo']

df.loc[players,'age':'club']
```

```python
#slice index lables
columns=['age','dob','height_cm','weight_kg']

#get top 1 top 10 player name
#df.index[:index]

df.loc['L. Messi':'M. Salah',columns]
```

Selecting with condition

```python
columns=['age','height_cm']
df.loc[df['height_cm']>180,columns]
```

```python
#multiple condition
df.loc[(df['height_cm']>180)&(df['nationality']=='Argentina'),:]
```
