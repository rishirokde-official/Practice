```python
#dictionary
dict_countries={
    'E0':'England',
    'D1':'Germany',
    'I1':'Ilaly',
    'F1':'France'

}
```

```python
dict_countries['F1']
```

```python
dict_historical_data={}

for league_code in dict_countries:
  frames_for_current_league = []
  for season in range (15,20):
    # Corrected: Use league_code (e.g., 'E0') directly as filename
    df=pd.read_csv(root + str(season)+ str(season+1)+ "/" + league_code + ".csv")
    df.insert(1,'season',season)
    frames_for_current_league.append(df)

  # Concatenate all seasons for the current league after the inner loop finishes
  df_current_league = pd.concat(frames_for_current_league, ignore_index=True)
  dict_historical_data[league_code] = df_current_league

# Concatenate all leagues into a master DataFrame after the outer loop finishes
df_master = pd.concat(dict_historical_data.values(), ignore_index=True)
df_master.to_csv("all_leagues_historical_data.csv",index=False)
```
