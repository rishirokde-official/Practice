```python
#create list
leagues=['P1','P2']

frames=[]
for league in leagues:
  df= pd.read_csv(root + "2122" + "/" + league + ".csv")
  frames.append(df)
```

```python
#MULTIPLE SEASONS
league_codes=['P1','P2']

frames=[]
#LOOP THROUGH RANGE OF SEASON AND INSERT NEW COLUMN
for league_code in league_codes :
  for season in range (15,21):
    season_str_part = str(season) + str(season+1)
    df=pd.read_csv(root + season_str_part + "/" + league_code + ".csv")
    df.insert(1,'season',season)
    frames.append(df)
```
