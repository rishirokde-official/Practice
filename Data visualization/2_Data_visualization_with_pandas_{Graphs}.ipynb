```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/rishirokde-official/Python-for-Data-Science/refs/heads/main/python-for-data-science-main/07.Project%202%20-%20Data%20Visualization/population_total.csv")
```

```python
df
```

```python
df.dropna(inplace=True)
```

```python
df_pivot=df.pivot(index='year',columns='country',values='population')
```

```python
df_pivot=df_pivot[['United States','India','China','Indonesia','Brazil']]
```

```python
df_pivot
```

Lineplot

```python
df_pivot.plot(kind='line',
         xlabel='Years',ylabel='Population',
         title='Population(1955-2020)',
         figsize=(10,6))
```

Barplot

Single Barplot

```python
#selecting only one year
df_pivot_2020=df_pivot[df_pivot.index.isin([2020])].copy()
```

```python
df_pivot_2020
```

```python
#making barplot
df_pivot_2020.plot(kind='bar',color='blue',
xlabel='years',ylabel='Population',
title='Population in 2020')
```

Barplot grouped by “n” variables

```python
#selecting few years
df_pivot_sample=df_pivot[df_pivot.index.isin([1980,1990,2000,2020])]
```

```python
# making grouped barplot
df_pivot_sample.plot(kind='bar',
                     xlabel='Years', ylabel='Population',
                     figsize=(10,4))
```
