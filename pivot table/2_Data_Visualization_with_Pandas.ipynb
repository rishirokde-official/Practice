```python
import pandas as pd
```

```python
df_population=pd.read_csv("https://raw.githubusercontent.com/rishirokde-official/Python-for-Data-Science/refs/heads/main/python-for-data-science-main/07.Project%202%20-%20Data%20Visualization/population_total.csv")
```

```python
df_population
```

```python
# dropping null values
df_population.dropna(inplace=True)
```

```python
# making a pivot table
df_pivot = df_population.pivot(index='year', columns='country',
                                        values='population')
```

```python
# selecting some countries
df_pivot = df_pivot[['United States', 'India', 'China',
                          'Indonesia', 'Brazil']]
```

```python
df_pivot
```

making visualizations Lineplot

```python
df_pivot.plot(kind='line',
              xlabel='Years',
              ylabel='population',title='population (1955-2020)',figsize=(8,4))
```

Barplot

```python
#want only 2020 year data
df_pivot_2020=df_pivot[df_pivot.index.isin([2020])]
```

```python
df_pivot_2020
```

```python
 #transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot_2020.T
```

```python
# making barplot
df_pivot_2020.plot(kind='bar', color='orange',
                   xlabel='Years', ylabel='Population',
                   title='Population in 2020')
```

Barplot grouped by “n” variables

```python
#want only few years data
df_pivot_sample=df_pivot[df_pivot.index.isin([1980,1990,2000,2010,2020])]
```

```python
# making grouped barplot
df_pivot_sample.plot(kind='bar',
                     xlabel='Years', ylabel='Population',
                     figsize=(8,4))
```

Piechart

```python
# changing column name
df_pivot_2020.rename(columns={2020:'2020'}, inplace=True)
```

```python
# making the plot
df_pivot_2020.plot(kind='pie',
                   y='2020',
                   title='Population in 2020 (%)')
```
