```python
import pandas as pd
```

```python
df_population_raw=pd.read_csv("https://raw.githubusercontent.com/rishirokde-official/Python-for-Data-Science/refs/heads/main/python-for-data-science-main/07.Project%202%20-%20Data%20Visualization/population_total.csv")
```

```python
df_population_raw
```

Making a Pivot Table


```python
#dropping null values (not)
df_population_raw.dropna(inplace=True)
```

```python
# making a pivot table
df_pivot=df_population_raw.pivot(index='year',columns='country',values='population')
```

```python
df_pivot
```

```python
#selecting countries
df_pivot=df_pivot[['United States','India','China','Indonesia','Brazil']]
```
